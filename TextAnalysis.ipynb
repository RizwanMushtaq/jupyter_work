```python
import string

PUNCTUATION = set(string.punctuation)
PUNCTUATION
```

```python
from collections import Counter

class TextAnalyzer:
    """ Text Analyser class to process text. """
    
    def __init__(self, text):
        self.text = self.remove_punctuation_set(text).lower()

    def remove_punctuation_set(self, s):
        return "".join(c for c in s if c not in PUNCTUATION)

    def freqAll(self):
        words = self.text.split()
        return Counter(words)
        

tA = TextAnalyzer("Hello world! How are you? I'm good, hope you are too!")
tA.text
```

```python
tA.freqAll()
```

```python
tA.freqAll().most_common(2)
```

```python
tA.freqAll()["good"]
```
