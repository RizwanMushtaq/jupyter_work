# Dictionaries

* denoted by {}
* Key and values pairs
* The keys must be immutable and unique
* The values can immutable, mutable and duplicates
* example -> dict1 = {"thriller": "1982", "horror": "1950"}
* using it -> dict1["thriller"], dict1.keys(), ...

```python
dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
dict["key1"]
```

```python
dict[(0,1)]
```

```python
dict.keys()
```

```python
dict.values()
```

```python
dict["key5"] = "hello"
dict
```

```python
list(dict.keys())
```

# Sets

* type of collection
* unlike lists and tuples, they are unordered
* They do not record element position
* sets only have unique elements
* define  a set -> set1 = {"pop", "rock", "disco"}
* convert list to set -> album_set = set(album_list)

```python
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
set1
```

```python
set1.add("hello")
set1
```

```python
set1.remove("hello")
set1
```
