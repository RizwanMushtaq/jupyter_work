```python
#To simplify file handling and ensure proper closure of files, 
#Python provides the "with" statement. 
#It automatically closes the file when operations 
#within the indented block are completed. 
#This is considered best practice when working with files.

with open("example1.txt", "r") as file:
    file_stuff = file.read()
print(file_stuff)
```

```python
import requests

filename = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBMDeveloperSkillsNetwork-PY0101EN-SkillsNetwork/labs/Module%204/data/example1.txt"


def download(url, save_as):
    # Fetch the file over standard HTTP
    response = requests.get(url)

    # Check if the download was successful (Status code 200)
    if response.status_code == 200:
        with open(save_as, "wb") as f:
            f.write(response.content)
        print("done")
    else:
        print(f"Failed to download. Status code: {response.status_code}")


# Run the function
download(filename, "example2.txt")
```

```python
# Create a new file Example2.txt for writing
with open('Example20.txt', 'w') as file20:
    file20.write("This is line A\n")
    file20.write("This is line B\n")
    # file1 is automatically closed when the 'with' block exits
```

```python
# List of lines to write to the file
Lines = ["This is line 1", "This is line 2", "This is line 3"]

# Create a new file Example3.txt for writing
with open('Example30.txt', 'w') as file30:
    for line in Lines:
        file30.write(line + "\n")
    # file2 is automatically closed when the 'with' block exits
```

```python
# Data to append to the existing file
new_data = "This is line C"

# Open an existing file Example2.txt for appending
with open('Example30.txt', 'a') as file30:
    file30.write(new_data + "\n")
    # file1 is automatically closed when the 'with' block exits
```

# Pandas: 
* Pandas is a popular open-source data manipulation and analysis library for the Python programming language
* It has functions for analyzing, cleaning, exploring, and manipulating data

Key features and functionalities of Pandas
Data Structures: Pandas offers two primary data structures - DataFrame and Series.

A DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

A Series is a one-dimensional labeled array, essentially a single column or row of data.
