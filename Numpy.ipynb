# Numpy

* NumPy stands for Numerical Python and it is an open source project.
* NumPy is a Python library used for working with arrays, linear algebra, fourier transform, and matrices.
* The array object in NumPy is called ndarray, it provides a lot of supporting functions that make working with ndarray very easy.
* NumPy is usually imported under the np alias.

* It's usually fixed in size and each element is of the same type. We can cast a list to a numpy array by first importing numpy.

# 1D Numpy

```python
import numpy as np
```

```python
# Create a numpy array

a = np.array([0, 1, 2, 3, 4])
a
```

# 1D Arrays: Vectors

* A 1D array is often termed as a vector.
* Depending upon the orientation of the data, the vector can be classified as a row vector or a column vector.
* Mathematically, we can add, subtract, and take the product of two vectors, provided they are the same shape.

# 2D Arrays : Matrices

* A 2D array is also called a Matrix.
* These are typically rectangular arrays with data stored in different rows.
* All of the operations mentioned above are also applicable to the 2D arrays.

```python
b = [[11, 12, 13], [21, 22, 23], [31, 32, 33]]
b
```

```python
# Convert list to Numpy Array
# Every element is the same type

B = np.array(b)
B
```

```python
# Show the numpy array dimensions

B.ndim
```

```python
# Show the numpy array shape

B.shape
```

```python
# Show the numpy array size

B.size
```

```python
# Access the element on the second row and third column

B[1, 2]
```

```python
# Access the element on the first row and first column

B[0][0]
```

```python
# Access the element on the first row and first and second columns

B[0][0:2]
```

```python
# Access the element on the first and second rows and third column

B[0:2, 2]
```
